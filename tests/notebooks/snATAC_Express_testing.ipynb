## Tutorial for implementing the snATAC-Express workflow

## 1. Set up environment and define input files

```bash
source ~/.bashrc
conda init bash
conda activate snatacexpress
```

```bash
# make the tiny test data the default input folder for every helper function
export SNATAC_EXPRESS_INPUT="$(pwd)/example_data/input_data"

# create an output folder for Phase-1
RESULT_DIR="$(pwd)/results/phase1"
mkdir -p "$RESULT_DIR"
echo "results will be written to: $RESULT_DIR"
```

## 2. Run Phase-1

```bash
# paths to the sparse matrices shipped with the repo
PEAK_MTX="example_data/input_data/sparse_peak_matrix.txt.mtx"
GEX_MTX="example_data/input_data/sparse_gex_matrix.txt.mtx"

python ../../scripts_phase1/run_multitest.py \
    -g    example_data/genelist_genebody.txt \
    -n    BACH2 \
    -gex  example_data/input_data/sparse_gex_matrix.txt.mtx \
    -pks  example_data/input_data/sparse_peak_matrix.txt.mtx \
    -pb   1 \
    -out  "$RESULT_DIR" | cat
```

```bash
# list everything the script produced
find "$RESULT_DIR" -type f | head -n 20
```
